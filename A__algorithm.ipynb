```python
import heapq

def a_star(graph, heuristic, start, goal):
    # Priority queue: (f_cost, g_cost, node, path)
    open_list = [(heuristic[start], 0, start, [start])]

    # Store the lowest cost found to each node
    g_costs = {start: 0}

    while open_list:
        f_cost, g_cost, current, path = heapq.heappop(open_list)

        # Goal reached
        if current == goal:
            return path, g_cost

        # Explore neighbors
        for neighbor, cost in graph[current].items():
            new_g_cost = g_cost + cost

            # If this path is better
            if neighbor not in g_costs or new_g_cost < g_costs[neighbor]:
                g_costs[neighbor] = new_g_cost

                # f(n) = g(n) + h(n)
                new_f_cost = new_g_cost + heuristic[neighbor]

                heapq.heappush(
                    open_list,
                    (new_f_cost, new_g_cost, neighbor, path + [neighbor])
                )

    return None, float('inf')


# Graph
graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'A': 1, 'C': 2, 'D': 5},
    'C': {'A': 4, 'B': 2, 'D': 1},
    'D': {'B': 5, 'C': 1, 'E': 3},
    'E': {'D': 3}
}

# Heuristic values: estimated cost from each node to E
heuristic = {
    'A': 7,
    'B': 6,
    'C': 2,
    'D': 3,
    'E': 0
}

# Find shortest path from A to E
path, cost = a_star(graph, heuristic, 'A', 'E')

if path:
    print("Path:", " -> ".join(path))
    print("Total Cost:", cost)
else:
    print("No path found")
```
