```python
import heapq # Import the heapq module for priority queue functionality

def uniform_cost_search(graph, start_node, goal_node):
    """
    Implements the Uniform Cost Search (UCS) algorithm.

    Args:
        graph (dict): A dictionary representing the graph where keys are nodes
                      and values are dictionaries of neighbors with their costs.
                      Example: {'A': {'B': 1, 'C': 4}, 'B': {'D': 2}, ...}
        start_node: The starting node for the search.
        goal_node: The target node to reach.

    Returns:
        tuple: A tuple containing:
            - list: The path from start_node to goal_node if found, otherwise None.
            - int: The total cost of the path if found, otherwise infinity.
    """

    # Priority queue stores (cost, current_node, path_to_current_node)
    # It is ordered by cost (smallest cost first)
    priority_queue = [(0, start_node, [start_node])]

    # Set to keep track of nodes that have been fully expanded (their neighbors processed)
    visited = set()

    # Dictionary to store the minimum cost found so far to reach each node
    cost_so_far = {start_node: 0}

    while priority_queue:
        # Pop the node with the smallest cost from the priority queue
        current_cost, current_node, current_path = heapq.heappop(priority_queue)

        # If the current node is the goal, we have found the optimal path
        if current_node == goal_node:
            return current_path, current_cost

        # If we have already visited this node with a lower or equal cost, skip it
        # This prevents redundant processing and ensures optimality
        if current_node in visited:
            continue

        # Mark the current node as visited
        visited.add(current_node)

        # Explore neighbors of the current node
        for neighbor, edge_cost in graph.get(current_node, {}).items():
            # Calculate the new cost to reach the neighbor through the current node
            new_cost = current_cost + edge_cost

            # If this new path to the neighbor is cheaper than any previously found path,
            # or if the neighbor hasn't been added to the priority queue yet
            if neighbor not in cost_so_far or new_cost < cost_so_far[neighbor]:
                cost_so_far[neighbor] = new_cost # Update the minimum cost to reach the neighbor
                new_path = current_path + [neighbor] # Extend the path
                heapq.heappush(priority_queue, (new_cost, neighbor, new_path)) # Add to priority queue

    # If the priority queue becomes empty and the goal was not reached, no path exists
    return None, float('inf')


# Example Usage:
if __name__ == "__main__":
    # Define a sample graph with more nodes and edge costs
    graph_1 = {
        'A': {'B': 1, 'C': 4},
        'B': {'D': 2, 'E': 5, 'F': 1},
        'C': {'F': 1, 'H': 7},
        'D': {'G': 3, 'I': 8},
        'E': {'G': 1, 'J': 2},
        'F': {'G': 2, 'K': 6},
        'G': {},
        'H': {'L': 3},
        'I': {'L': 1},
        'J': {'M': 4},
        'K': {'M': 1},
        'L': {'N': 2},
        'M': {'N': 5},
        'N': {}
    }

    start_1 = 'A'
    goal_1 = 'N'

    # Run UCS to find the path and cost
    path, cost = uniform_cost_search(graph_1, start_1, goal_1)

    if path:
        print(f"Path from {start_1} to {goal_1}: {path}")
        print(f"Total cost: {cost}")
    else:
        print(f"No path found from {start_1} to {goal_1}")
```
